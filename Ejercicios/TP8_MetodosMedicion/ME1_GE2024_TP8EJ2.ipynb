```python
# Datos

V2 = 10
R1 = 1001.1
R2 = 10000.1
N = 5
Vi_mean = 1e-3
Vi_std = 0.1e-3

err_R1 = 0.01 # en %
err_R2 = 0.01 # en %
err_volt_porc = 0.8
err_volt_cuentas = 1

rango_volt = 400e-3
cuenta_max = 4000

porcentaje_confianza_objetivo = 95.45
```

```python
# Calculo de la tensión Vx

Vx = V2 * R1/(R1+R2)
print("El valor de la tensión Vx es: {:.4f} V".format(Vx))
```

```python
# Calculo de la tensión V1

V1_mean = Vx + Vi_mean
print("El valor medio de la tensión medida V1 es: {:.4f} V".format(V1_mean))
```

```python
# Análisis de Incertidumbre Tipo B de Vx
# Tengo que hace la incertidumbre combinada de Vx, que seria la de V2 (que no tiene), y la de las resistencias.

ui_R1 = 0
ui_R2 = 0
uj_R1 = err_R1 / (100 * np.sqrt(3))
uj_R2 = err_R2 / (100 * np.sqrt(3))

uc_R1 = np.sqrt(ui_R1**2 + uj_R1**2)
uc_R2 = np.sqrt(ui_R2**2 + uj_R2**2)

dVx_dR1 = V2 * R2/((R1+R2)**2)
dVx_dR2 = -1 * V2 + R1/((R1+R2)**2)

uc_Vx = np.sqrt((dVx_dR1**2 * (uc_R1*R1)**2) + (dVx_dR2**2 * (uc_R2*R2)**2))
```

```python
# Análsis de Incertidumbre Tipo A y B de V1
ui_Vi = Vi_std
uj_Vi = (Vi_mean * err_volt_porc/100 + err_volt_cuentas * (rango_volt/cuenta_max))/np.sqrt(3)
uc_Vi = np.sqrt(ui_Vi**2 + uj_Vi**2)

uc_V1 = np.sqrt(uc_Vi**2 + uc_Vx**2)
print(uc_Vx)
```

```python
v_eff = uc_V1 ** 4 / (ui_Vi**4 / np.sqrt(N))
print(v_eff)
```

```python
K = ((stats.t.ppf((1.0+(porcentaje_confianza_objetivo/100))/2.0, v_eff)))
print(K)
```

```python
print("La tensión de la fuente V1 es: (", V1_mean, "+/- ", K*uc_V1, ") V @95%")
```
