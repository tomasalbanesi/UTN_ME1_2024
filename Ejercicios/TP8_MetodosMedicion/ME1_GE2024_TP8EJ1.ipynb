```python
# Datos

V_Rx_med = 9.143
V_Rx_std = 25.2e-3
V_Rx_n = 9

V_Rp_med = 5.097
V_Rp_std = 27.6e-3
V_Rp_n = 16

Rp = 10e6
U_Rp = 0.10e6
K_Rp = 2

V_Rx_v = V_Rx_n - 1
V_Rp_v = V_Rp_n - 1

err_V_Rx_lec = 0.05
err_V_Rx_cuentas = (2 / 9143) * 100

err_V_Rp_lec = 0.05
err_V_Rp_cuentas = (2 / 5097) * 100
```

```python
# Calculo de la resistencia Rx

Rx = (V_Rx_med / V_Rp_med) * Rp
print("El valor de la resistencia Rx es: {:.4f} M\u03A9".format(Rx/1e6))
```

```python
# Análisis de Incertidumbre Tipo A
ui_V_Rx = V_Rx_std / np.sqrt(V_Rx_n)
ui_V_Rp = V_Rp_std / np.sqrt(V_Rp_n)

print("El valor de la Incertiumbre Tipo A de V_Rx es: {:.4f}".format(ui_V_Rx))
print("El valor de la Incertiumbre Tipo A de V_Rp es: {:.4f}".format(ui_V_Rp))
```

```python
# Análsis de Incertidumbre Tipo B
uj_V_Rx = ((err_V_Rx_lec + err_V_Rx_cuentas) / 100) / np.sqrt(3) * V_Rx_med
uj_V_Rp = ((err_V_Rp_lec + err_V_Rp_cuentas) / 100) / np.sqrt(3) * V_Rp_med

print("El valor de la Incertiumbre Tipo B de V_Rx es: {:.4f}".format(uj_V_Rx))
print("El valor de la Incertiumbre Tipo B de V_Rp es: {:.4f}".format(uj_V_Rp))

uj_Rp = U_Rp/K_Rp
print("El valor de la Incertiumbre Tipo B de Rp es: {:.2f}".format(uj_Rp))
```

```python
# Incertidumbre combinada de Rp, V_Rx, V_Rp
uc_Rp = uj_Rp
uc_V_Rx = np.sqrt(ui_V_Rx**2 + uj_V_Rx**2)
uc_V_Rp = np.sqrt(ui_V_Rp**2 + ui_V_Rp**2)

# Incertidumbre combinada de Rx
dRx_dRp = V_Rx_med / V_Rp_med
dRx_dVRx = Rp / V_Rp_med
dRx_dVRp = - (V_Rx_med * Rp) / V_Rp_med**2

uc_Rx = np.sqrt( (dRx_dRp * uc_Rp)**2 + (dRx_dVRx * uc_V_Rx)**2 + (dRx_dVRp * uc_V_Rp)**2 )

print("El valor de la Incertiumbre Combinada de Rp es: {:.4f}".format(uc_Rp))
print("El valor de la Incertiumbre Combinada de V_Rp es: {:.4f}".format(uc_V_Rp))
print("El valor de la Incertiumbre Combinada de V_Rx es: {:.4f}".format(uc_V_Rx))
print("El valor de la Incertiumbre Combinada de Rx es: {:.4f}".format(uc_Rx))
```
